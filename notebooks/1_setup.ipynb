## Clean the dataset

```python
# Load the dataset csv file
df = pd.read_csv(f"{DATA_DIR}/styles.csv", on_bad_lines='skip')
df.head()
```

### Fix the missing images path

```python
df["images_path"] = DATA_DIR + "/images/" + df['id'].astype(str) + ".jpg"
```

```python
missing_images = df[~df['images_path'].apply(os.path.exists)]
print(f"Number of missing images: {len(missing_images)}")
```
