Load the dataset

```python
df = pd.read_csv(f"{DATA_DIR}/cleaned_products.csv")
df.head()
```

### Clean the text

```python
def clean_text(text):
  text = text.lower() # Convert the text into lowercase
  text = re.sub(r"[^a-z0-9\s]", "", text) # Remove punctuations and special sybols
  text = re.sub(r"\s+", " ", text).strip() # Remove extra whitespace

  return text
```

Add metadata to the input text

```python
df['text_input'] = (
    df['productDisplayName'].fillna('') + '. ' +
    "Category : " + df['masterCategory'] + '. ' +
    "Type : " + df['articleType'] + '. ' +
    "Color: " + df['baseColour'].fillna('')
    )
```

```python
# Remove the punctuations, special symbols, and whitespaces
df['text_input'] = df['text_input'].apply(clean_text)
```
